# 03 Visualisierung

Dieses Notebook erstellt eine einfache Grafik aus künstlichen Portfolio-Daten.

```python
import pandas as pd
import matplotlib.pyplot as plt

portfolio = pd.DataFrame({
    "Jahr": [2020, 2021, 2022, 2023, 2024, 2025],
    "Portfolio": [100, 113, 105, 122, 136, 151],
    "Vergleichsindex": [100, 110, 101, 118, 130, 142],
})

ax = portfolio.plot(x="Jahr", y=["Portfolio", "Vergleichsindex"], marker="o")
ax.set_title("Beispielhafte Portfolio-Entwicklung")
ax.set_xlabel("Jahr")
ax.set_ylabel("Indexwert")
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
