<h2>6.1 Beantwortung der Forschungsfrage</h2>

<p>Die Forschungsfrage dieser Arbeit lautet: «Wie können historische Finanzdaten mithilfe eines Python-basierten Modells verwendet werden, um langfristige Investmentstrategien hinsichtlich Risiko und Rendite zu vergleichen?»</p>

<p>Die Untersuchung zeigt, dass ein historischer Vergleich möglich ist, wenn Daten vereinheitlicht und Anlageregeln vorab festgelegt werden. Das Modell zeigt neben Endvermögen die Schwankungen und zwischenzeitlichen Verluste; Gewichte und Signale machen Entscheidungen nachvollziehbar.</p>

<p>Dafür ist die ganze Daten-, Rechen- und Kontrollkette entscheidend. Historische BIP-Entscheidungen benötigen damals verfügbare Datenstände. Dokumentierte Eingaben, Einstellungen und Ergebnisse ermöglichen Wiederholung und Prüfung einzelner Schritte. Damit wurde ein nachvollziehbares Werkzeug für den Rendite-Risiko-Vergleich erstellt und praktisch angewendet, jedoch keine Vorhersage künftiger Renditen.</p>

<h2>6.2 Zentrale Erkenntnisse aus dem Vergleich</h2>

<p>Aus den Ergebnissen in Abschnitt 5.4 und ihrer Interpretation in Abschnitt 5.5 ergibt sich keine Strategie, die bei allen betrachteten Kriterien überlegen ist. Das ist für mich die wichtigste inhaltliche Erkenntnis: Eine höhere Rendite bedeutet nicht automatisch, dass eine Strategie insgesamt überzeugender abschneidet.</p>

<p>Buy-and-Hold mit dem breiten Aktien-ETF erzielte sowohl im globalen Hauptvergleich als auch im längeren US-Vergleich mehr Endvermögen als das jeweilige 60/40-Portfolio und die Trendfolge mit den festgelegten Hauptparametern. Dieser Ertrag ging jedoch mit stärkeren Schwankungen als beim 60/40-Portfolio einher. Besonders deutlich zeigt sich das im US-Lauf: Trotz des höchsten Endvermögens fiel das Buy-and-Hold-Portfolio zwischenzeitlich um rund 51 % vom vorherigen Höchststand. Ein Blick auf den Endwert allein würde diese zwischenzeitlichen Verluste ausblenden.</p>

<p>Das jährlich angepasste 60/40-Portfolio erreichte in beiden Untersuchungsblöcken die höchste Sharpe Ratio und eine niedrigere Volatilität als reines Aktien-Buy-and-Hold. Bezogen auf diese Kriterien war es damit der günstigere Kompromiss, allerdings bei geringerem Ertrag und weiterhin möglichen Verlusten. Das Ergebnis gilt für die Kombination aus Aktien, Staatsanleihen und jährlichem Rebalancing. Welchen Anteil das Rebalancing allein daran hatte, wurde nicht gesondert untersucht.</p>

<p>Bei der Trendfolge fällt das Urteil weniger einheitlich aus. Im globalen Hauptlauf erzielte sie weniger Rendite und einen grösseren maximalen Drawdown als Buy-and-Hold. Im längeren US-Lauf war ihr grösster Rückgang dagegen mit rund 22 % deutlich kleiner. Eine Regel zum Ein- und Aussteigen bietet somit in den untersuchten Daten keinen durchgehend besseren Verlustschutz. Die ergänzenden SMA-Varianten zeigen ausserdem, dass bereits andere Fensterlängen das Ergebnis deutlich verändern können. Da sich zwischen dem globalen und dem US-Lauf sowohl Markt als auch Zeitraum ändern, lässt sich deren jeweiliger Einfluss nicht getrennt bestimmen.</p>

<p>Die BIP-Gewichtung lag im Hauptlauf beim Endvermögen knapp vor dem Welt-ETF. Dieser Vergleich allein belegt aber keinen Vorteil der Gewichtungsregel, weil unterschiedliche Märkte und Fonds enthalten sind. Aussagekräftiger ist der Kontrolllauf mit denselben sieben Länder-ETFs: Dort erreichte die BIP-Variante mehr Rendite bei etwas geringerer Volatilität als die Gleichgewichtung. Ihr deutlich höherer US-Anteil und die starke Entwicklung des US-Fonds machen den Unterschied nachvollziehbar. Daraus folgt weder, dass eine grosse Volkswirtschaft automatisch hohe Aktienrenditen liefert, noch dass BIP-Gewichtung allgemein besser als Marktkapitalisierungsgewichtung ist. Letztere wurde im Kontrolllauf nicht untersucht.</p>

<p>Für die Beurteilung langfristiger Strategien reicht es deshalb nicht, nur nach dem höchsten historischen Ertrag zu fragen. Ebenso wichtig ist, welche Rückgänge eine Strategie mit sich bringt und nach welchem Risikomass sie beurteilt wird. In dieser Untersuchung führen diese unterschiedlichen Blickwinkel zu unterschiedlichen Rangfolgen.</p>

<h2>6.3 Grenzen der Schlussfolgerungen</h2>

<p>Die Ergebnisse gelten für die ausgewählten Fonds, Zeiträume und Regeln. Auch die bestandenen Tests und Gegenrechnungen machen daraus keinen Beweis für eine künftige Überlegenheit. Sie zeigen innerhalb ihres Prüfumfangs, dass die Berechnungen zu den festgehaltenen Eingaben passen. Ob die ursprünglichen Finanzdaten vollständig richtig sind und wie gut die vereinfachten Annahmen die Wirklichkeit abbilden, ist davon zu unterscheiden.</p>

<p>Die Grenzen aus Abschnitt 5.5.6 bleiben für das Fazit entscheidend: Zusätzliche Transaktionskosten, persönliche Steuern und Inflation fehlen, enthaltene Fondskosten bleiben berücksichtigt. Cash ist unverzinst, die Sharpe-Vergleichsrendite angenähert. Monatswerte zeigen keine vollständigen Verluste innerhalb des Monats. Die Ergebnisse sind deshalb weder persönliche Nettorenditen noch eine vollständige Risikodarstellung.</p>

<p>Gerade kleine Abstände, wie zwischen dem Welt-ETF und dem BIP-Portfolio, sollten deshalb nicht überbewertet werden. Es wurde nicht statistisch nachgewiesen, dass solche Unterschiede über die betrachtete Stichprobe hinaus verlässlich bestehen.</p>

<h2>6.4 Rückblick auf die Arbeit und Ausblick</h2>

<p>Für mich liegt der Beitrag der Arbeit in der Verbindung von Finanztheorie und Informatik. Aus den einzelnen theoretischen Bausteinen entstand ein System, mit dem sich konkrete Anlageideen untersuchen lassen. Dabei war nicht nur das Zusammenführen der Formeln wichtig. Ebenso entscheidend waren eindeutige Daten, zeitlich passende Entscheidungen und Kontrollen, die Fehler sichtbar machen.</p>

<p>Bei der technischen Umsetzung und bei den Prüfungen kam KI-Unterstützung zum Einsatz, wie in den Arbeitsnachweisen dokumentiert. Die Berechnungen wurden anhand der festgelegten Regeln und der verwendeten Daten überprüft. Dadurch konnte ich sicherstellen, dass die Ergebnisse innerhalb des geprüften Umfangs nachvollziehbar und rechnerisch korrekt sind. Die Verantwortung für die Auswahl der Untersuchungsmethoden und die Interpretation der Ergebnisse liegt weiterhin bei mir. Dabei muss ich besonders darauf achten, dass meine Schlussfolgerungen durch die tatsächlichen Ergebnisse gedeckt sind und nicht über den untersuchten Zeitraum hinaus verallgemeinert werden.</p>

<p>Eine Weiterführung könnte gezielter einzelne Einflüsse untersuchen, etwa durch den Vergleich desselben 60/40-Portfolios mit und ohne jährliche Umschichtung. Auch Transaktionskosten oder verzinste Cash-Phasen wären sinnvolle Ergänzungen. Eine spätere Weboberfläche könnte das vorhandene Modell leichter zugänglich machen; für die hier durchgeführte Untersuchung ist sie nicht erforderlich.</p>

<p>Insgesamt liefert die Arbeit damit keine allgemeine Kaufempfehlung, sondern eine überprüfbare Grundlage für den Vergleich historischer Anlagestrategien. Ihr Nutzen liegt darin, nicht nur zu zeigen, welche Strategie mehr verdient hat, sondern auch, welche Risiken und Annahmen hinter diesem Ergebnis stehen.</p>
