Die folgenden Abschnitte unterscheiden geplante Eingangskontrollen von den dokumentierten Prüfungen und Nachweisen der durchgeführten Untersuchung. Die Qualitätsprüfung betrifft Daten und Berechnungsregeln; sie ist kein Nachweis einer künftig sicheren Rendite. Die wichtigsten Angaben zur Durchführung stehen in Abschnitt 5.3, die Grenzen der Aussagekraft in 5.5.6.


## Prüfregeln vor der Simulation {#sec-pruefung-vor-simulation}



### Validierung vor der Simulation {#sec-pruefung-5-5-2}



<p>Bevor ein Datensatz in die Simulation gelangt, soll er automatisch geprüft werden. Dabei sind mindestens folgende Punkte relevant:</p>

- die benötigten Spalten sind vorhanden,
- Datumswerte können korrekt eingelesen werden,
- Wertspalten sind numerisch,
- <code>asset_id</code> ist vorhanden,
- die Kombination aus Datum und Anlage kommt nicht doppelt vor,
- fehlende oder unendliche Werte werden erkannt,
- der verfügbare Zeitraum wird für jede Anlage bestimmt.

<p>Fehlende Werte sind bei realen Datensätzen grundsätzlich möglich und werden von pandas ausdrücklich als eigener Bestandteil der Datenaufbereitung behandelt [@mckinneyPythonDataAnalysis2022, Kap. 7]. Für diese Simulation sollen solche Stellen deshalb sichtbar gemacht und nicht unbemerkt weiterverarbeitet werden.</p>

<p>Das Ergebnis der Prüfung wird zusätzlich in einem kleinen Datenbericht gespeichert. So lässt sich später nachvollziehen, ob eine Simulation mit vollständigen Daten oder mit Warnungen durchgeführt wurde.</p>


### Umgang mit fehlenden Marktwerten {#sec-pruefung-5-5-3}



<p>Fehlende Marktwerte werden in der ersten Version nicht automatisch linear interpoliert. Eine solche Interpolation würde einen Wert erzeugen, der in der ursprünglichen Datenquelle nicht beobachtet wurde. Dadurch könnten auch künstliche Renditen oder Signale entstehen.</p>
<p>Nicht jede fehlende Kalenderbeobachtung ist allerdings ein Datenfehler. An Wochenenden oder Feiertagen findet beispielsweise kein regulärer Handel statt. Deshalb soll die Prüfung zwischen einem fehlenden Handelstag und einer tatsächlichen Lücke in der vorhandenen Reihe unterscheiden können.</p>
<p>Wenn mehrere Anlagen gemeinsam verwendet werden, muss zudem festgelegt werden, wie unterschiedliche Handelstage ausgerichtet werden. Die genaue Regel wird mit den später ausgewählten Datensätzen getestet. Wichtig ist bereits jetzt, dass kein Wert ohne dokumentierte Regel ergänzt wird.</p>

## Eingangsprüfung der WEO-Dateien {#sec-weo-eingang}


<p>Für die BIP-Daten wurden die 16 Herbstausgaben des World Economic Outlook (WEO) aus den Jahren 2009 bis 2024 heruntergeladen und jeweils in einem eigenen Ordner gespeichert. Für jede Datei wurden die Quelle, das Download-Datum und der ursprüngliche Dateiname dokumentiert. Anschliessend wurde überprüft, ob alle Ausgaben die benötigten BIP-Daten der sieben untersuchten Länder enthalten und ob zwischen historischen Werten und Schätzungen unterschieden wird. Dabei wurden auch Unterschiede zwischen den Dateiformaten sowie Besonderheiten bei der Auswahl der historischen Jahre festgehalten, um diese bei der späteren Datenaufbereitung berücksichtigen zu können.</p>

## Ausführungsnachweise und Gegenrechnungen {#sec-run-nachweise}


<p>Ein Aufruf bestand aus dem Start der Engine mit der jeweiligen Konfigurationsdatei, beispielsweise <code>configs/study_v1/H1.json</code>. Diese Dateien verweisen auf die archivierten Datenkopien, nicht auf veränderliche Arbeitsdateien. Es wurden keine neuen Parameter gesucht und keine weiteren Strategien hinzugefügt.</p>

<p>Für jeden Lauf entstand ein eigener Ordner unter <code>outputs/runs/</code>. <code>portfolio_history.csv</code> enthält den monatlichen Vermögensverlauf, <code>summary.csv</code> die zusammengefassten Kennzahlen. Je nach Strategie kommen Gewichte, Umschichtungen und Signale hinzu. Das Run-Manifest verbindet diese Ergebnisse mit den verwendeten Eingaben, der Konfiguration und den Prüfsummen. Die vollständigen Run-Kennungen und Zeitstempel bleiben in den technischen Nachweisen erhalten, statt den Fliesstext mit langen Identifikationsnummern zu belasten.</p>

<p>Anschliessend erfolgte eine zusätzliche KI-gestützte Gegenrechnung aus den eingefrorenen Eingaben und den Ergebnisdateien. Dabei wurden unter anderem Portfoliowerte, Renditen, Risikokennzahlen, Signalverzögerungen und jährliche Umschichtungen geprüft. Innerhalb dieses dokumentierten Prüfumfangs wurden keine rechnerischen Abweichungen gefunden. Die fünf Ergebnisordner bleiben unverändert erhalten. Die folgenden Tabellen und Grafiken werden daraus abgeleitet; für ihre Darstellung werden keine neuen Hauptläufe ausgeführt.</p>
