<h2>4.1 Untersuchungsansatz</h2>

<p>Ich untersuche die Anlagestrategien mit einem historischen Backtest. Dabei wird berechnet, wie sich ein Portfolio entwickelt hätte, wenn festgelegte Regeln auf vergangene Finanzdaten angewendet worden wären. Mich interessiert sowohl der erzielte Ertrag als auch der Verlauf dorthin. Dafür verbinde ich die Finanzgrundlagen aus Kapitel 3 mit einem Python-Modell.</p>

<p>Die Arbeit trennt das allgemeine Rechensystem von der konkreten Untersuchung. Das System soll geeignete Daten und unterschiedliche Einstellungen verarbeiten können. Für die Maturarbeit lege ich dagegen bestimmte Anlagen, Zeiträume und Regeln vor der Auswertung fest. So verändere ich nicht nachträglich die Einstellungen, nur weil eine andere Variante besser abschneidet. Hilpisch beschreibt diese Gefahr als Überanpassung: Eine auf einem Datensatz besonders erfolgreiche Einstellung muss auf anderen Daten nicht gleich gut funktionieren [@hilpischPythonFinanceMastering2019, Kapitel 15, «Optimization»].</p>

<p>Dieses Kapitel fasst die Vergleichsregeln zusammen. Ihre ausführliche Herleitung, die technische Umsetzung und die Durchführung stehen in Kapitel 5. Die verwendeten Daten, Einstellungen und Programmstände werden dort nachvollziehbar zugeordnet. Die KI-Unterstützung bei Umsetzung und Prüfungen wird in den Arbeitsnachweisen und im Anhang dokumentiert.</p>

<h2>4.2 Untersuchte Anlagestrategien</h2>

<h3>4.2.1 Buy-and-Hold</h3>

<p>Bei Buy-and-Hold bleibt das Kapital während des gesamten Untersuchungszeitraums in einem breiten Aktien-ETF investiert. Es gibt keine regelmässigen Ein- und Ausstiege. Im Hauptvergleich verwende ich dafür VT als weltweiten Aktienfonds, im längeren US-Vergleich VTI. Buy-and-Hold dient innerhalb des jeweiligen Vergleichs als einfache Referenz. Die konkrete Fondsauswahl wird in Abschnitt 5.10.2 begründet.</p>

<h3>4.2.2 60/40-Portfolio mit jährlichem Rebalancing</h3>

<p>Dieses Portfolio beginnt mit 60 % im jeweiligen Aktien-ETF und 40 % im Staatsanleihen-ETF IEF. Zwischen den Umschichtungen verändern sich die Anteile durch die unterschiedliche Wertentwicklung. Am Jahresende wird die Verteilung wieder auf 60/40 gesetzt. Die neuen Positionen bestimmen die Rendite der folgenden Periode; am letzten Untersuchungsdatum erfolgt keine erneute Umschichtung.</p>

<p>IEF bildet US-Staatsanleihen mit Restlaufzeiten von sieben bis zehn Jahren ab [@IShares710Year]. Der Anleihenanteil ist damit kein risikofreies Guthaben. Untersucht wird die Kombination aus Aktien, Anleihen und jährlicher Anpassung. Der Nutzen des Rebalancings allein wird nicht isoliert gemessen, weil kein ansonsten identisches Portfolio ohne Umschichtungen als eigener Lauf vorgesehen ist.</p>

<h3>4.2.3 Trendfolge mit gleitenden Durchschnitten</h3>

<p>Die Trendstrategie vergleicht einen kurzen mit einem langen gleitenden Durchschnitt derselben bereinigten Aktienwertreihe. Diese Durchschnitte werden als SMA bezeichnet. Die Hauptvariante verwendet drei beziehungsweise zwölf Monatsbeobachtungen. Liegt der kurze Durchschnitt über dem langen, wird in den Aktien-ETF investiert. Andernfalls bleibt das Kapital in Cash, das im Modell unverzinst ist. Auf fallende Kurse wird nicht gesetzt.</p>

<p>Ein Signal vom Monatsende gilt erst für den nächsten Monat. Es darf nicht rückwirkend die Rendite des gerade abgelaufenen Monats verändern. Bereits vor dem Untersuchungsbeginn müssen genügend Beobachtungen für beide Durchschnitte vorliegen. Dieser Vorlauf dient nur der Signalberechnung und zählt nicht zur ausgewerteten Portfoliorendite. Die Varianten 2/12 und 6/12 ergänzen den Vergleich, ohne die Hauptregel nachträglich zu ersetzen.</p>

<h3>4.2.4 BIP-Gewichtung ausgewählter Länder</h3>

<p>Die Ländergewichtung greift das von Kommer erläuterte Konzept auf, Länderanteile anhand ihrer Wirtschaftsleistung festzulegen [@kommerSouveraenInvestierenMit2025, Abschnitt 9.4]. Für meine Untersuchung verwende ich die USA, Kanada, Japan, das Vereinigte Königreich, Deutschland, Frankreich und Italien. Jedes Land wird durch einen Aktien-ETF vertreten. Sein Zielanteil ergibt sich aus seinem nominalen, in US-Dollar angegebenen BIP im Verhältnis zur BIP-Summe dieser sieben Länder.</p>

<p>Die Zielgewichte werden zu Beginn und an den jährlichen Umschichtungsterminen neu bestimmt. Alle Länder werden anhand desselben BIP-Jahres gewichtet. Ich verwende das neueste gemeinsame Jahr mit Werten, die damals bereits als historische Angaben vorlagen. Schätzungen und Prognosen sind ausgeschlossen. Gibt es mehrere Versionen, zählt für jedes Land die zuletzt beim Entscheid verfügbare. Spätere Korrekturen werden nicht rückwirkend eingesetzt. Zwischen den jährlichen Anpassungen verändern sich die tatsächlichen Anteile durch die Fondsrenditen. Dies ist eine reine BIP-Ländergewichtung, keine Nachbildung des Gerd-Kommer-ETFs mit dessen zusätzlichen Gewichtungs- und Faktorregeln.</p>

<h2>4.3 Vergleichsbedingungen und Daten</h2>

<p>Die Untersuchung besteht aus fünf festgelegten Läufen. Die Kürzel dienen ihrer eindeutigen Zuordnung; sie bezeichnen keine zusätzlichen Finanzmethoden.</p>

| Lauf | Renditezeitraum | Vergleich |
|:---|:---|:---|
| **H1** | 2010–2025 | VT Buy-and-Hold, VT/IEF 60/40, VT-Trend 3/12 und BIP-Länderportfolio |
| **K1** | 2010–2025 | BIP-Gewichtung gegenüber jährlicher Gleichgewichtung derselben sieben Länder-ETFs |
| **Z1** | 2003–2025 | VTI Buy-and-Hold, VTI/IEF 60/40 und VTI-Trend 3/12 |
| **S1** | 2010–2025 | VT-Trend 2/12 |
| **S2** | 2010–2025 | VT-Trend 6/12 |

<p>Im Hauptvergleich stehen unterschiedliche Anlagekonzepte gegenüber. Weil VT und das Länderportfolio nicht dieselben Märkte enthalten, lässt sich ein Unterschied nicht allein auf die BIP-Regel zurückführen. K1 hält deshalb die sieben Länderfonds gleich und verändert nur ihre Gewichtung: nach BIP oder mit jeweils einem Siebtel. Das ist kein Vergleich mit Marktkapitalisierungsgewichten. Z1 ergänzt die Untersuchung um eine längere US-Stichprobe; hier ändern sich gegenüber H1 sowohl Markt als auch Zeitraum.</p>

<p>Jedes Portfolio startet mit 100’000 USD, ohne weitere Einzahlungen oder Entnahmen. Innerhalb eines Laufs verwenden alle Strategien dieselben monatlichen Bewertungstermine. Für die Renditen ab Januar 2010 beziehungsweise Januar 2003 wird zusätzlich die Anfangsbewertung Ende Dezember des Vorjahres benötigt. Die Wahl der Zeiträume und der notwendige SMA-Vorlauf werden in Abschnitt 5.10 erläutert.</p>

<p>Die täglichen ETF-Daten beziehe ich über <code>yfinance</code> von Yahoo Finance. Ausgewertet wird <code>Adj Close</code>, das Yahoo als um Dividenden und Splits bereinigten Schlusskurs beschreibt [@WhatAdjustedClose]. Dividenden werden deshalb nicht nochmals hinzuaddiert. Für jeden Monat wird die letzte tatsächlich vorhandene Bewertung übernommen; ihr ursprüngliches Datum bleibt dokumentiert. Fehlende notwendige Werte werden nicht still aufgefüllt. Produktangaben von Vanguard und iShares ergänzen die Prüfung, welche Anlagen die Fonds vertreten.</p>

<p>FRED liefert mit <code>DGS3MO</code> die Zinsreihe für den risikofreien Vergleich [@MarketYieldUS2026]. Aus einer vor Periodenbeginn bekannten Jahreszinsnotierung wird eine monatliche Vergleichsrendite angenähert. Sie beeinflusst die Sharpe Ratio, nicht die unverzinste Cash-Position. Die genaue Umrechnung wird in Abschnitt 5.10.7 erläutert.</p>

<p>Die BIP-Werte stammen aus den archivierten WEO-Herbstausgaben des Internationalen Währungsfonds (IMF) von 2009 bis 2024. So lassen sich Bezugsjahr und damaliger Datenstand unterscheiden. Die Einstiegsseiten der ältesten und jüngsten verwendeten Ausgabe sind dokumentiert [@DownloadEntireWorld; @WorldEconomicOutlook]. Die Regeln zur historischen Verfügbarkeit stehen in Abschnitt 5.10.8.</p>

<h2>4.4 Auswertung und Grenzen</h2>

<p>Für den Ertrag betrachte ich Endvermögen, Gesamtrendite und annualisierte geometrische Rendite. Für das Risiko verwende ich Volatilität und maximalen Drawdown. Letzterer beschreibt den grössten beobachteten Rückgang vom bisherigen Höchststand, nicht eine Grenze für mögliche künftige Verluste. Für die Sharpe Ratio ziehe ich von jeder Monatsrendite die risikofreie Vergleichsrendite ab. Der Durchschnitt dieser Überschussrenditen wird durch ihre Standardabweichung geteilt und auf Jahresbasis umgerechnet. Für die Annualisierung werden zwölf Monatsperioden pro Jahr angesetzt.</p>

<p>Ich beurteile die Strategien nicht anhand einer einzigen Rangliste. Kommer erläutert, dass unterschiedliche Risikokennzahlen unterschiedliche Seiten des Anlagerisikos erfassen [@kommerSouveraenInvestierenMit2025, Abschnitt 2.2]. Neben den Kennzahlen betrachte ich deshalb die Vermögensverläufe und ziehe Gewichte, Umschichtungen und Signale zur Erklärung heran. Ergänzende Teilperioden werden aus den durchgehenden Verläufen ausgewertet, ohne die Portfolios neu zu starten.</p>

<p>Die Ergebnisse werden weder um Inflation bereinigt noch in Schweizer Franken umgerechnet. Persönliche Steuern und zusätzliche Transaktionskosten werden nicht simuliert; bereits in den ETF-Reihen enthaltene Fondskosten bleiben dagegen Bestandteil der Entwicklung. Die monatliche Beobachtung erfasst nicht alle zwischenzeitlichen Verluste. Auch die Ausführung zu den verwendeten Schlussbewertungen ist eine Vereinfachung.</p>

<p>Daten, Einstellungen und Programmstand werden vor den Hauptläufen festgehalten. Tests und Gegenrechnungen prüfen, ob die Berechnungen den vorgegebenen Regeln entsprechen. Die Schlussfolgerungen bleiben auf die untersuchten Daten und Annahmen begrenzt; eine statistisch nachgewiesene oder künftig sichere Überlegenheit einer Strategie wird damit nicht beansprucht.</p>
