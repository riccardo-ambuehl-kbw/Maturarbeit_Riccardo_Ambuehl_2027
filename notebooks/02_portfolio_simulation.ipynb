# 02 Portfolio-Simulation

Dieses Notebook zeigt eine einfache Portfolio-Berechnung mit festen Gewichtungen.

```python
import pandas as pd

renditen = pd.DataFrame({
    "ETF_Global": [0.12, -0.05, 0.14, 0.10, 0.09],
    "Aktie_A": [0.25, -0.27, 0.51, 0.09, 0.15],
    "Anleihe_B": [0.03, -0.02, 0.04, 0.03, 0.03],
})

gewichte = pd.Series({
    "ETF_Global": 0.60,
    "Aktie_A": 0.25,
    "Anleihe_B": 0.15,
})

portfolio_rendite = renditen.dot(gewichte)
portfolio_wert = (1 + portfolio_rendite).cumprod() * 100

pd.DataFrame({
    "Portfolio_Rendite": portfolio_rendite,
    "Portfolio_Wert": portfolio_wert,
})
```

```python
kennzahlen = pd.Series({
        "Durchschnittliche Rendite": portfolio_rendite.mean(),
        "Volatilitaet": portfolio_rendite.std(),
        "Endwert": portfolio_wert.iloc[-1],
    })

kennzahlen
```
